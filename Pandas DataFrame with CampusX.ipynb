```python
points_df = pd.DataFrame(
    {
        '1st point':[(2,3),(6,5),(0,0),(7,8),(-1,9)],
        '2nd point':[(2,6),(5,7),(8,-4),(4,9),(6,7)]
    }
)
points_df
```

```python
def euclidean(row):
    pt_A=row['1st point']
    pt_B=row['2nd point']

    return ((pt_A[0]-pt_B[0])**2 + (pt_A[1]-pt_B[1])**2)**0.5
```

```python
points_distance = points_df.apply(euclidean,axis = 1)
points_distance 
```
